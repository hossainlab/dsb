```python
Map  Functions
```

```python
def square(x):
    return x*x
```

```python
num =  [1,2,3,4,5]
```

```python
list(map(square,num))
```

```python
Filter Functions
```

```python
num1 =  [1,2,3,4,5]
```

```python
result = list(filter(lambda x :x%2==0,num1))
```
