## 5.2.1 Visulaizing Distributions of Data
- **What is distributions**
  - A distribution describes how the values of a dataset are spread or arranged.
  - Common types of distributions include normal, skewed, uniform, and multimodal distributions.

- **Key Terms:**
    - **Mean :** The average value
    - **Median :** The middle value when data is sorted.
    - **Mode:** The most frequently occurring value.
    - **Variance and Standard Deviation:** Measures of data spread.

- **Common Plot Types:**
    - Histograms
    - Boxplots
    - Violin plots

### `fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1)`

This line creates a figure with a grid of subplots and assigns each subplot to a specific variable. Here's a breakdown of the components:

- `fig`: The `Figure` object that represents the entire plotting area or "canvas" containing all subplots.
- `(ax1, ax2)`: Tuple of `Axes` objects, where each `Axes` object represents an individual subplot.

#### Explanation of `nrows=2, ncols=1`

- `nrows=2`: Specifies that the grid has **2 rows**.
- `ncols=1`: Specifies that the grid has **1 column**.

This configuration creates a figure with **two stacked subplots** (one above the other).

- `ax1`: Refers to the **first subplot** (top row).
- `ax2`: Refers to the **second subplot** (bottom row).

You can use `ax1` and `ax2` to customize each subplot independently, like adding titles, labels, or specific plot types.


```python
# separate 2007's data
df_2007 = df[df.year == 2007]
```

```python
# separate Asia and Europe's data
asia_2007 = df_2007[df_2007.continent == "Asia"]
europe_2007 = df_2007[df_2007.continent == "Europe"]
```

```python
plt.hist(asia_2007.gdpPercap)
plt.show() 
```

```python
plt.hist(europe_2007.gdpPercap)
plt.show() 
```

```python
plt.subplot(211)
plt.title('Distribution of GDP Percapita')
plt.hist(asia_2007.gdpPercap, edgecolor = 'black', bins = 20)
plt.ylabel('Asia')

plt.subplot(212)
plt.hist(europe_2007.gdpPercap, edgecolor = 'black', bins = 20)
plt.ylabel('Europe')
plt.show() 
```

### `plt.subplot`

`plt.subplot` is used to add a subplot to the current figure. It divides the figure into a grid of subplots, specified by the parameters.

### `plt.subplot(211)`

The argument `211` breaks down as follows:
- The first digit (`2`) represents the number of **rows** in the grid.
- The second digit (`1`) represents the number of **columns** in the grid.
- The third digit (`1`) selects the **first subplot** in the grid.

So, `plt.subplot(211)` creates a grid with **2 rows and 1 column**, and `211` refers to the first plot in that 2x1 grid.


```python
fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1)
ax1.hist(asia_2007.gdpPercap, edgecolor = 'black', bins = 20)
ax2.hist(europe_2007.gdpPercap, edgecolor = 'black', bins = 20)

ax1.set_xlabel('GDP', fontsize=16)
ax1.set_ylabel('Asia', fontsize=16)

ax2.set_xlabel('GDP', fontsize=16)
ax2.set_ylabel('Europe', fontsize=16)

plt.tight_layout()
plt.savefig("GDP.png") # .pdf, .png, .jpg, .jepg, .tiff

plt.show() 
```

### Boxplots

### Boxplot Components

#### Box
- Represents the **interquartile range (IQR)**, which is the range between the **first quartile (Q1, 25th percentile)** and **third quartile (Q3, 75th percentile)**.
- The box height (or width, in horizontal boxplots) indicates the spread of the middle 50% of the data.

#### Median Line
- A line inside the box shows the **median** (50th percentile) of the dataset.
- The position of this line within the box can indicate skewness; if it’s closer to Q1, the data is **right-skewed**.

#### Whiskers
- Lines, called **whiskers**, extend from the box to the smallest and largest values within a specified range, usually up to **1.5 times the IQR** from Q1 and Q3.
- The whiskers give a general idea of the **data's range**, excluding outliers.

#### Outliers
- Individual points beyond the whiskers are marked as **outliers** and are typically plotted as dots or asterisks.
- Outliers are values that fall outside **1.5 times the IQR** from Q1 or Q3.

## 5.2.3 Visualizing distributions with Seaborn

```python
# conventional import of seaborn
import seaborn as sns
```

### Histogram

```python
sns.histplot(df['Age'], kde=True, bins=20)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution with KDE')
plt.show()
```

```python
sns.set(style='whitegrid')
sns.set_palette('viridis')
sns.histplot(df['Age'], kde=True, bins=20)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution with KDE')
plt.show()
```

- **KDE**
  - A kernel density estimate (KDE) plot is a method for visualizing the distribution of observations in a dataset, analogous to a histogram.
  - KDE represents the data using a continuous probability density curve in one or more dimensions

```python
# divide plot by a category
sns.set(font_scale=2, palette= "viridis")
sns.displot(data=df, x="Age", col="Gender")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.set(font_scale=2, palette= "viridis")
sns.histplot(data=df, x="Age", bins=20, hue='Gender') #categorized on gender column
plt.show()
```

### Boxplots

```python
sns.boxplot(x='Gender', y='Age', data=df)
plt.title('Boxplot of Age by Gender')
plt.show()
```

### Violinplot
- To show the distribution and density of the data.

```python
sns.violinplot(x='Gender', y='Age', data=df)
plt.title('Violin Plot of Age by Gender')
plt.show()
```
