## 5.3.1 Introduction
- **Importance of Categorical Data Visualization:** Categorical data represents types of data that can be divided into groups or categories. Visualizing these data helps in understanding relationships and differences between categories.

## 5.3.2 Understanding Categorical Data
- **Definition:** Categorical data consists of values that represent distinct groups or categories. Examples include gender, color, type of vehicle, or survey responses (e.g., yes/no).

- **Types of Categorical Data:**

    - **Nominal:** Categories with no intrinsic order (e.g., gender, color).
    - **Ordinal:** Categories with a specific order (e.g., ratings like poor, fair, good).
- **Common Plot Types for Categorical Data:**
    - Bar plots
    - Pie charts
    - Count plots

## 5.3.3 Visualizing Categorical Data with Matplotlib

### Barplots

```python
# import libraries
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_excel('../Data/LungCapData.xlsx')
df
```

```python
# Aggregate data: calculate the mean age for each gender
LungCap_by_gender = df.groupby('Gender')['LungCap'].mean()
```

```python
LungCap_by_gender
```

```python
# Plotting the bar chart
plt.bar(LungCap_by_gender.index, LungCap_by_gender, color='skyblue')
plt.xlabel('Gender')
plt.ylabel('Average LungCap')
plt.title('Average LungCap by Gender')
plt.show()
```

```python
labels = ['Category A', 'Category B', 'Category C', 'Category D']
sizes = [20, 30, 25, 25]  # Percentage or count of each category
colors = ['lightblue', 'lightgreen', 'lightcoral', 'gold']  # Optional: colors for each slice

# Plotting the pie chart
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors)
plt.title('Example Pie Chart')
plt.show()
```

**Explanation of the Chart**
- Each slice of the pie chart represents the proportion of total height contributed by each gender.
- `Labels` show each gender.
- `autopct='%1.1f%%'` displays the percentage for each slice.
- `startangle=140` rotates the pie chart for a better visual alignment.

```python
labels = ['Category A', 'Category B', 'Category C', 'Category D']
sizes = [20, 30, 25, 25]  
colors = ['lightblue', 'lightgreen', 'lightcoral', 'gold']  
explode = [0, 0.1, 0, 0]  # "Explode" the second slice (Category B)

# Plotting the pie chart
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors, explode=explode)
plt.title('Example Pie Chart')
plt.show()
```

## 5.3.4 Visualizing Categorical Data with Seaborn

- **Count Plots**
    - **Purpose:** To visualize the count of observations in each category.

```python
sns.countplot(x=df['Gender'], data=df, palette='viridis')
plt.title('Count Plot of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```
