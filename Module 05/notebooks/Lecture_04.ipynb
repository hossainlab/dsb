## 5.4.1 Introduction
- **Importance of Visualizing Statistical Relationships:** Visualizing relationships between variables is crucial for understanding data patterns, correlations, and insights that might not be immediately apparent through numerical analysis alone.

## 5.4.2 Understanding Statistical Relationships
- **Definition:** A statistical relationship refers to the connection or association between two or more variables.

- **Types of Relationships:**

    - **Positive Correlation:** As one variable increases, the other also increases.
    - **Negative Correlation:** As one variable increases, the other decreases.
    - **No Correlation:** There is no discernible relationship between the variables.
- **Key Metrics:**

    - **Correlation Coefficient (r):** A numerical measure that indicates the strength and direction of a linear relationship between two variables (range from -1 to 1).
    - **p-value:** Used to determine the significance of the relationship; a low p-value indicates a statistically significant relationship.
- **Common plot types:**

    - **Scatter plots**
    - **Line charts**
    - **Pair plots**
    - **Heatmaps**

## 5.4.3 Visualizing Relationships with Matplotlib
### Scatterplots

```python
#conventional import
import matplotlib.pyplot as plt
```

```python
# Sample Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue', alpha=0.7)
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.title('Scatter Plot of X vs Y')
plt.grid(True)
plt.show()
```

### Line plots

```python
plt.plot(x, y, marker='o', linestyle='-', color='green')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.title('Line Plot of X vs Y')
plt.grid(True)
plt.show()
```

```python
# Sample data
years = [2018, 2019, 2020, 2021, 2022]
values1 = [100, 220, 290, 330, 500]  # Data for the first line
values2 = [50, 120, 200, 350, 550]  # Data for the second line

# Plotting both lines on the same figure
plt.figure(figsize=(10, 6))
plt.plot(years, values1, label='Line 1', color='blue', marker='o', linestyle='-')
plt.plot(years, values2, label='Line 2', color='green',marker='x', linestyle='--')

# Adding labels and title
plt.xlabel('Year')
plt.ylabel('Values')
plt.title('Two Line Charts in the Same Figure')
plt.legend()  # Show legend to differentiate lines

# Display the plot
plt.show()
```

## 5.4.4 Visualizing Relationships with Seaborn
### Scatter plot with Regression line

```python
import seaborn as sns
import pandas as pd
```

```python
# Sample DataFrame
df = pd.DataFrame({
    'X': [1, 2, 3, 4, 5],
    'Y': [2, 3, 5, 7, 11]
})

sns.regplot(x='X', y='Y', data=df, scatter_kws={'color':'blue'}, line_kws={'color':'red'})
plt.title('Scatter Plot with Regression Line')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.show()
```

## Pair plot

```python
df = pd.read_excel('../Data/LungCapData.xlsx')
df
```

```python
sns.pairplot(data=df,  palette='viridis')
plt.title('Pair Plot of Iris Dataset')
plt.show()
```

### Heatmaps

```python
numerical_df = df[['Age', 'Height','LungCap']]
```

```python
correlation = numerical_df.corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

- A value of 1 indicates a perfect positive correlation.
- A value of 0 indicates no correlation.
- A value of -1 indicates a perfect negative correlation.
- **Create the Heatmap:**

    `sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')`
This line uses Seaborn to plot a heatmap of the correlation matrix.
- **Parameters:**
    - **correlation:** The data (correlation matrix) to display.
    - **annot=True:** Displays correlation values on the heatmap.
    - **cmap='coolwarm':** Sets the color map from cool (negative correlations) to warm (positive correlations).
    - **fmt='.2f':** Formats each correlation value to two decimal places.
