- **Independent Variable**: A variable which value does not change by the effect of other variables and is used to manipulate the dependent variable. It often denoted as $X$.

- **Dependent Variable**: A variable whose value change when there is any manipulation in the values of independent variable. It is often denoted as $Y$

**Example**: Salary depends on year of experience

- **Salary**: Dependent Variable 
- **Year of experience**: Independent Variable

## <img src="https://img.icons8.com/?size=100&id=61864&format=png&color=000000" style="height:50px;display:inline"> What is Machine Learning?
---
* Machine learning is a subfield of AI, and is a program or system that trains a model from input data. 
* Once the model has been trained, it can then make meaningful predictions on new, unseen data similar to what it was trained on. 
* Machine learning gives the computer the ability to **learn without explicit programming**.

One definition: *"Machine learning is the semi-automated extraction of knowledge from data"*

- **Knowledge from data**: Starts with a question that might be answerable using data
- **Automated extraction**: A computer provides the insight
- **Semi-automated**: Requires many smart decisions by a human
---
Two of the most common classes of machine learning models are **unsupervised** and **supervised**. The key difference between the two is that supervised models have labels. Labeled data is data that comes with a tag, like a name,a type, or a number. Labels are similar to column headers in a spreadsheet. Unlabeled data does not have a tag.

## <img src="https://img.icons8.com/?size=100&id=d0mvB0b7Ml0f&format=png&color=000000" style="height:50px;display:inline"> Supervised Learning 
- The majority of practical machine learning uses supervised learning.
- Supervised learning is where you have input variables ($X$) and an output variable ($Y$) and you use an algorithm to learn the mapping function from the input to the output.
$$Y = f(X)$$
- The goal is to approximate the mapping function so well that when you have new input data (X) that you can predict the output variables (Y) for that data.
- It is called supervised learning because the process of an algorithm learning from the training dataset can be thought of as a teacher supervising the learning process. We know the correct answers, the algorithm iteratively makes predictions on the training data and is corrected by the teacher. Learning stops when the algorithm achieves an acceptable level of performance.

---
Supervised learning problems can be further categorized into regression and classification tasks.

* **Classification**: Classification is a process of finding a function which helps in dividing the dataset into classes based on different parameters. In Classification, a computer program is trained on the training dataset and based on that training, it categorizes the data into different classes.

* **Regression**: Regression is a process of finding the correlations between dependent and independent variables.

<center><img src="../img/supervised-ml.png"/></center>

## <img src="https://img.icons8.com/?size=100&id=aIClzC3iqUok&format=png&color=000000" style="height:50px;display:inline"> Unsupervised Learning

- Unsupervised learning is where you only have input data ($X$) and no corresponding output variables.
- The goal for unsupervised learning is to model the underlying structure or distribution in the data in order to learn more about the data.
- These are called unsupervised learning because unlike supervised learning above there is no correct answers and there is no teacher. Algorithms are left to their own devises to discover and present the interesting structure in the data.

---
Unsupervised learning problems can be further grouped into clustering and association problems.
- **Clustering**: A clustering problem is where you want to discover the inherent groupings in the data, such as grouping customers by purchasing behavior.
- **Association**:  An association rule learning problem is where you want to discover rules that describe large portions of your data, such as people that buy $X$ also tend to buy $Y$.

## <img src="https://img.icons8.com/?size=100&id=fTkqveCX0blI&format=png&color=000000" style="height:50px;display:inline"> Traditional Machine Learning 
---

- **Input Components:**
  - **Training Code:** The set of instructions and algorithms used to train the model.
  - **Labeled Data:** Data that includes input-output pairs for supervised learning.

- **Processing:**
  - **Model Building:** Combines the training code and labeled data to develop a machine learning model.

- **Output Predictions:**
  - The trained model can perform one of the following tasks:
    - **Predict:** Make numerical forecasts (Regression).
    - **Classify:** Assign data points to specific categories (Classification).
    - **Cluster:** Group similar data points together without predefined labels (Clustering).

## <img src="https://img.icons8.com/?size=100&id=kTuxVYRKeKEY&format=png&color=000000" style="height:50px;display:inline"> Generative AI 
---

- **Input Components:**
  - **Training Code:** The set of instructions and algorithms used to train the foundation model.
  - **Labeled Data:** Data with predefined categories or annotations, used for supervised learning.
  - **Unlabeled Data:** Raw data without predefined categories, allowing the model to learn patterns independently.

- **Processing:**
  - **Foundation Model:** A large-scale AI model trained on vast amounts of data, capable of generalizing across various tasks.

- **Output Generation:**
  - The trained foundation model generates new content based on inputs, which can be applied in:
    - **Text Generation:** AI-powered tools for generating human-like text (e.g., ChatGPT, GPT-4).
    - **Code Generation:** Automated code completion and suggestion models (e.g., GitHub Copilot).
    - **Image Generation:** AI models capable of creating original images from textual descriptions (e.g., DALL·E, Stable Diffusion).

## <img src="https://img.icons8.com/?size=100&id=HkG28tSEJLgP&format=png&color=000000" style="height:50px;display:inline"> Understanding Machine Learning vs. Generative AI
---
- **ML System Workflow:**
  - Takes **input** from a dataset.
  - Processes it using a **Machine Learning (ML) system**.
  - Produces an **output** based on the model's learning.

- **When It Is Not Generative AI (Gen AI):**
  - If the output (**y**) is:
    - A **Number** (e.g., predicting sales figures).
    - A **Class** (e.g., classifying emails as spam or not spam).
    - A **Probability** (e.g., likelihood of an event occurring).

- **When It Is Generative AI (Gen AI):**
  - If the output (**y**) is:
    - **Natural Language** (e.g., text generation like ChatGPT).
    - **Image** (e.g., AI-generated art like DALL·E).
    - **Audio** (e.g., AI-generated music or speech synthesis).

---

1. **Evolution of AI**  
   - Artificial Intelligence (AI) has progressed significantly since its inception in the 1950s, evolving from general AI concepts to more specialized fields such as machine learning, deep learning, and generative AI.

2. **Machine Learning (1980s)**  
   - The development of machine learning in the 1980s enabled AI systems to learn from historical data, moving beyond rule-based programming to statistical modeling and pattern recognition.

3. **Deep Learning (2010s)**  
   - The emergence of deep learning in the 2010s introduced neural network-based models that mimic human brain function, leading to major advancements in image recognition, natural language processing, and autonomous systems.

4. **Generative AI (2020s)**  
   - The latest advancement in AI, Generative AI (Gen AI), leverages deep learning foundation models to create original content, revolutionizing fields such as text generation, image synthesis, and creative automation.
