### 4.3.1 Selecting Specific Columns

```python
# conventional import
import pandas as pd
```

```python
# reading dataset
df = pd.read_csv("../data/diabetes.csv")
df
```

```python
# Select 'Glucose', 'Insulin' and 'Age' columns
subset_df = df[['Glucose', 'Age', 'Insulin']]
subset_df
```

### 4.3.2 Selecting Columns with `loc[]`

```python
# Select all rows but only 'Name' and 'Salary' columns
subset_df = df.loc[:, ['Glucose', 'BloodPressure', 'Age']]
subset_df
```

### 4.3.3 Selecting Columns by Index Position with `iloc[]`

```python
# Select the first two columns
subset_df = df.iloc[:, :2]
subset_df
```

```python
# Select columns from position 1 to 3
subset_df = df.iloc[:, 1:4]
subset_df
```

### 4.3.4 Selecting Columns Based on Data Type
Use the `.select_dtypes()` method to select columns of a specific data type

```python
# Select only numeric columns
subset_df = df.select_dtypes(include='number')
subset_df
```

```python
# Select only float columns
subset_df = df.select_dtypes(include='float')
subset_df
```

### 4.3.5 Selecting Columns by Pattern (Regex)
Use `.filter()` to select columns matching a pattern

```python
# Select columns that start with 'Ski'
subset_df = df.filter(regex='^Ski')
subset_df
```

```python
# Select columns that end with 'es'
subset_df = df.filter(regex='es$')
subset_df
```

```python
# Select all columns where the column name contains the letter "B" anywhere in the name
subset_df = df.filter(like='B')
subset_df
```

### 4.3.6 Selecting Columns Using a Boolean Mask

```python
# Select columns where the mean value is greater than 50
subset_df = df.loc[:, df.mean() > 50]
subset_df
```

```python
# mean value for each column
df.mean()
```

### 4.3.8 Selecting Columns Based on Column Name Length

```python
# Select columns where the column name has more than 5 characters
subset_df = df.loc[:, df.columns.str.len() > 5]
subset_df
```

### 4.3.9 Selecting Columns by Condition Applied to Row Values
You can use `.any()` or `.all()` to select columns based on values in rows

```python
# Select columns where any value is greater than 100
subset_df = df.loc[:, (df > 100).any()]
subset_df
```

```python
# Select columns where all values are not null
subset_df = df.loc[:, df.notnull().all()]
subset_df
```
