```python
#conventional import
import pandas as pd
```

```python
# reading dataset
df = pd.read_csv("../data/LungCapData.csv")
df
```

## 4.4.1 Selecting Specific Rows and Columns

```python
# Select rows where Age > 10 and include only 'Age','LungCap', 'Height', 'Gender' columns
subset_df = df.loc[df['Age'] > 10, ['Age','LungCap', 'Height', 'Gender']]
subset_df
```

### 4.4.2 Creating New Columns

### 4.4.2.1 Create a column with a constant value

```python
# Add a new column with a constant value of 0
df['NewColumn'] = 0
df
```

### 4.4.2.2 Using `.assign()` to Add a New Column

```python
# Add a column using assign; can also chain this with other operations
df = df.assign(Height_cm=df['Height'] * 2.54)
df
```

```python
# Filter rows where 'BMI' is greater than equal 23 and add a new column with a calculated value
subset_df = df[df['Gender'] == 'male'].assign(height_in_meter=df['Height_cm'] / 100)
subset_df
```

### 4.4.2.3 Using Existing Columns to Create a Calculated Column

```python
# Add a column that is the product of two existing columns
df['height_multiply_age'] = df['Height'] * df['Age']
df
```

### 4.4.2.4 Adding a Column Based on Conditions with `.apply()` and `lambda`

```python
# Add a new column based on conditions
df['Caesarean_Number'] = df['Caesarean'].apply(lambda x: 1 if x == 'yes' else 0)
df
```

### 4.4.2.5 Create Columns Using `.loc`

```python
# Add column based on a condition
df.loc[df['Age'] > 19, 'Status'] = 'Non Teen'
df.loc[df['Age'] > 12 & (df['Age'] <= 19), 'Status'] = 'Teen'
df.loc[df['Age'] < 12, 'Status'] = 'Child'
df
```

### 4.4.2.6 Create Columns using `.query()`

```python
subset_df = df.query('Age < 15').assign(need_care=lambda x: x['Smoke'] == 'yes')
subset_df
```

**Explanation**
1. **`df.query('Age < 15')`**: This filters the rows in `df` where the column `Age` is less than 15, creating a subset of `df`.

2. **`.assign(need_care=lambda x: x['Smoke'] == 'yes')`**: The `.assign()` method adds a new column, `need_care`, to the subset. It uses a lambda function that checks if the value in the `Smoke` column is `'yes'`. If `Smoke` is `'yes'`, `need_care` will be `True`; otherwise, it will be `False`.

The resulting DataFrame, `subset_df`, contains only rows where `Age` is less than 15 and includes an additional `need_care` column based on whether `Smoke` is `'yes'`.


### 4.4.2.7 Create Columns using `.isin()`

```python
covid_df = pd.read_csv("../data/covid19.csv")
covid_df
```

```python
# Select rows where Country/Region is either 'Ukrain' or 'Netherlands', add a "death_affected_ratio" column
subset_df = covid_df[covid_df['Country/Region'].isin(['Ukraine', 'Netherlands'])].assign(death_affected_ratio=lambda x: x['Deaths'] / x['Confirmed'])
subset_df
                                                                                  
```

### 4.4.3 Drop specific columns

```python
# Select rows where Deaths > 100, keep all columns except "Recovered" and "Last Update"
subset_df = covid_df.loc[covid_df['Deaths'] > 100].drop(columns=['Recovered', 'Last Update'])
subset_df
```
