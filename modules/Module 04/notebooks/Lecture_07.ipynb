### 4.7.1 Understanding GroupBy objects
---

The "group by" concept: we want to **apply the same function on subsets of your dataframe, based on some key to split the dataframe in subsets**

This operation is also referred to as the "split-apply-combine" operation, involving the following steps:

* **Splitting** the data into groups based on some criteria
* **Applying** a function to each group independently
* **Combining** the results into a data structure

```python
df = pd.read_csv("../data/covid19.csv")
df.head()
```

```python
df.info()
```

```python
# drop null values
df.dropna(inplace=True)
```

![](../img/group_by_cities.png)

```python
# group by "Country/Region"
country = df.groupby("Country/Region")
```

```python
print(list(country))
```

```python
# get number of unique group
country.nunique()
```

```python
# count rows for each groups
country.count()
```

```python
# get each group's max value
country.max()
```

```python
# get min for each group
country.min()
```

```python
# describe all statistics for each group
country.describe()
```

```python
#  get data for individual group
country.get_group("Brazil")
```

### 4.7.3 Split-apply-combine

![](../img/split_apply_combine.png)
**This method of splitting your dataset in smaller groups and then applying an operation 
(such as min or max) to get aggregate result is called Split-Apply-Combine. It is illustrated in a diagram below**

```python
df = pd.read_csv("../data/LungCapData.csv")
df
```

```python
# group by "Gender" then find average of age in each group
average_age_by_gender = df.groupby('Gender')["Age"].mean()
average_age_by_gender
```

```python
# group by gender and find max of "LungCap" and "Age" column
max_lungCap_age_each_gender = df.groupby("Gender")[["LungCap", "Age"]].max() 
max_lungCap_age_each_gender
```

### 4.7.4 Advanced Aggregation with `agg()`

```python
df = pd.read_table("../data/pulse.txt")
```

```python
df.head()
```

```python
group_by_exercise_and_average_all_columns = df.groupby("Exercise").mean()
group_by_exercise_and_average_all_columns
```

```python
group_by_exercise_find_all_statistics = df.groupby("Gender").agg(["min", "max", "mean", "median", "std"])
group_by_exercise_find_all_statistics
```

```python
# group by "gender" find basic statistics on "Height" and "Weight" column
group_gender_find_stat_on_height_weight =  df.groupby("Gender").agg({"Height": ["min", "max", "mean", "median", "std"],
                                          "Weight": ["min", "max", "mean", "median", "std"]})
group_gender_find_stat_on_height_weight
```
