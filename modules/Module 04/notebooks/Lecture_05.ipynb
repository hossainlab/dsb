### 4.5.3 Replacing Missing Values with `fillna()`

2. **`round(..., 1)`**: The mean value is rounded to one decimal place using the `round()` function.

The output shows the average age of passengers on the Titanic, rounded to one decimal place.


```python
# fill the null value of 'Age' column with 'mean'
df.Age.fillna(mean, inplace = True)
```

```python
df.isna().sum()
```

```python
# get median of age column and round it to 1 decimal place
median = df.Age.median()
median = round(median, 1)
median
```

```python
# fill the null value of 'Age' column with 'median'
df.Age.fillna(median, inplace = True)
```

### 4.5.5 Outliers Detection and Removal
An `outlier` is a data point significantly different from other observations in a dataset. Outliers can occur due to variability in the data, errors in data collection, or unexpected values and can skew statistical analysis, leading to misleading results

#### 4.5.5.1 Outliers Detection and Removal using `IQR`

```python
df = pd.read_table("../data/pulse.txt")
df.head()
```

- **Q1 (First Quartile)**: The 25th percentile, or the "lower quartile."
  - Represents the value below which 25% of the data fall.

- **Q2 (Second Quartile)**: The 50th percentile, also known as the **median**.
  - Represents the value below which 50% of the data fall.

- **Q3 (Third Quartile)**: The 75th percentile, or the "upper quartile."
  - Represents the value below which 75% of the data fall.

These quartiles are used in calculating the **Interquartile Range (IQR)**, which is often applied to detect **outliers**:

**IQR = Q3 - Q1**

Values below **( Q1 - 1.5*IQR )** or above **( Q3 + 1.5 *IQR )** are typically considered outliers.


```python
# calculate quantile 
Q1, Q2, Q3 = df['Height'].quantile([.25, .50, .75])
print("Q1 25 percentile of the given data is: ", Q1)
print("Q2 50 percentile of the given data is: ", Q2)
print("Q3 75 percentile of the given data is: ", Q3)
```

```python
# find range
r = df.Height.max() - df.Height.min() 
print(r)
```

```python
# interquartile range (iqr)
IQR = Q3 - Q1 
print(IQR)
```

```python
# set upper and lower limit [Q1 - 1.5 x IQR, Q3 + 1.5 x IQR]
lower = Q1 - 1.5 * IQR 
upper = Q3 + 1.5 * IQR 
lower, upper
```

```python
df.shape
```

```python
# detect & removal outliers 
df_new = df[(df['Height'] < upper) & (df['Height'] > lower)]
df_new
```

```python
df_new.shape
```

#### 4.5.5.2 Outliers Detection and Removal using `Percentile`

```python
df = pd.read_table("../data/pulse.txt")
```

```python
df.head()
```

```python
# set maximum thresold 
max_thresold = df['Height'].quantile(0.95)
max_thresold
```

```python
# detect outliers 
df[df['Height'] > max_thresold]
```

```python
# set minimum thresold 
min_thresold = df['Height'].quantile(0.05)
min_thresold
```

```python
# detect outliers 
df[df['Height'] < min_thresold]
```

```python
# remove outliers 
df_clean = df[(df['Height'] < max_thresold) & (df['Height'] > min_thresold)]
df_clean
```

```python
# shape of dataset 
df.shape, df_clean.shape
```

#### 4.5.5.3 Outliers Detection and Removal using `z-score`

$$
Z = \frac{x - \mu}{\sigma}
$$

```python
df['ZScore'] = (df.Height - df.Height.mean())/df.Height.std() 
df.head() 
```

```python
# detect 
df[(df['ZScore'] < -3) | df['ZScore'] > 3]
```

```python
# removal 
df_new = df[(df['ZScore'] > -3) | df['ZScore'] < 3]
df_new
```
