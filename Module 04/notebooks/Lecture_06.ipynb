### 4.6.2 Descriptive Statistics

```python
data = {
    'Age': [25, 35, 45, 32, 40, 38],
    'Salary': [50000, 80000, 70000, 72000, 68000, 75000],
    'Profession': ["Business", "Engineer", "Engineer", "Business", "Doctor", "Engineer"]
}
```

```python
df = pd.DataFrame(data)
```

```python
# basic statistics on each column
df.describe()
```

```python
# individual column statistics
df['Age'].mean()
```

### 4.6.3 Summarizing Categorical Data

```python
# Count occurrences of each category in a column
df['Profession'].value_counts()
```

```python
pulse_df['Age'].value_counts()
```

```python
# to get proportions instead of raw counts
pulse_df['Age'].value_counts(normalize = True)
```
