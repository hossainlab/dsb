### 4.8.1 Changing Layout

 ### **`pivot()` and `pivot_table()`**

- **`pivot()`**: Reshape data based on column values.
  - **Syntax**: `df.pivot(index='index_column', columns='column_to_pivot', values='values')`

- **`pivot_table()`**: Similar to `pivot()`, but allows for aggregation with duplicate entries.


```python
#conventional import
import pandas as pd
```

```python
# Sample DataFrame
data = {
    'Date': ['2024-01', '2024-01', '2024-02', '2024-02'],
    'Category': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 250]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Pivot table to view monthly sales by category
df.pivot(index='Date', columns='Category', values='Sales')
```

### **`melt()`**
Converts wide format to long format.
- **Syntax**: `df.melt(id_vars=['fixed_columns'], value_vars=['columns_to_melt'])`

```python
# Melting Sales by Category
df.melt(id_vars='Date', value_vars='Sales')
```

- **Objective**: Transform the `Sales` column to a "long" format, with each `Sales` value associated with a `Date`.
  
- **Explanation**:
  - `id_vars='Date'`: Keeps the `Date` column as is (as the identifier variable).
  - `value_vars='Sales'`: Transforms the `Sales` column into two new columns:
    - `variable`: Contains the original column name ("Sales") repeated for each row.
    - `value`: Contains the actual values from the `Sales` column.

- **Result**: The data is "melted" into a long format with `Date` and `Sales` values in separate rows.


### 4.8.2 Sorting Data
- Sorting by Columns with `.sort_values()`
- Sort data based on values in one or more columns.
- **`Syntax: df.sort_values(by=['column_name'], ascending=True)`**

```python
# Sort data by sales in descending order
df.sort_values(by='Sales', ascending=False)
```

```python
# Sort data by sales in ascending order
df.sort_values(by='Sales', ascending=True)
```

### 4.8.3 Reindexing Data
- Reindex Rows
- Use `.reindex()` to change the index labels or add missing ones.
- **`Syntax: df.reindex(new_index)`**

```python
import numpy as np
 
column=['a','b','c','d','e']
index=['A','B','C','D','E']
 
# create a dataframe of random values of array
df1 = pd.DataFrame(np.random.randint(100, size=(5,5)), 
            columns=column, index=index)
 
df1
```

```python
# after reindexing
df1.reindex(['B', 'D', 'A', 'C', 'E'])
```

```python
# after reindexing
df1.reindex(['X', 'D', 'A', 'C', 'R'])
```

### Fill Missing Values in Reindexing
- Use `.reindex(fill_value=0)` to add default values for new rows.

```python
df1.reindex(['X','A', 'U'], fill_value=0)
```

### 4.8.4 Renaming Rows and Columns
- Renaming Columns
- Use `.rename()` to rename columns, either with a dictionary or using keyword arguments.
- **`Syntax: df.rename(columns={'old_name': 'new_name'})`**

```python
df
```

```python
# Renaming 'Sales' column to 'Revenue'
df.rename(columns={'Sales': 'Revenue'}, inplace=True)
```

```python
df
```

- Renaming Rows (Index)
- Use `.rename()` to rename index labels.

```python
# Renaming index row 'A' to 'a'
df1.rename(index={'A': 'a'})
```
