### 4.9.1 Merging Dataframes
![](../img/db_joins.jpg)

```python
#conventional import
import pandas as pd
```

```python
staff_df = pd.DataFrame([{'Name': 'Kelly', 'Role': 'Director of HR'},
                         {'Name': 'Sally', 'Role': 'Course liasion'},
                         {'Name': 'James', 'Role': 'Grader'}])
staff_df = staff_df.set_index('Name')
staff_df
```

```python
student_df = pd.DataFrame([{'Name': 'James', 'School': 'Business'},
                           {'Name': 'Mike', 'School': 'Law'},
                           {'Name': 'Sally', 'School': 'Engineering'}])
student_df = student_df.set_index('Name')
student_df
```

### 4.9.1.1 Outer Join
FULL (OUTER) JOIN: Returns all records when there is a match in either left or right table
![](../img/img_fulljoin.gif)

```python
pd.merge(staff_df, student_df, how='outer', on = 'Name')
```

### Explanation:

The `pd.merge()` function merges two DataFrames, `staff_df` and `student_df`, based on their indexes.

- `staff_df, student_df`:  
  These are the two DataFrames to be merged.

- `how='outer'`:  
  Specifies an outer join, meaning all rows from both DataFrames are included, even if they don't have a match in the other DataFrame. Missing values are filled with `NaN`.

- `on = 'Name'`:  
  Indicates that the merge should be done based on the `Name` column

This creates a new DataFrame with rows from both `staff_df` and `student_df`, where rows with the same index are combined, and rows with no corresponding match in the other DataFrame have `NaN` values in the resulting DataFrame.


### 4.9.2 Inner Join
(INNER) JOIN: Returns records that have matching values in both tables
![](../img/img_innerjoin.gif)

```python
pd.merge(staff_df, student_df, how='inner', on = 'Name')
```

### 4.9.3 Left Join
LEFT (OUTER) JOIN: Returns all records from the left table, and the matched records from the right table
![](../img/img_leftjoin.gif)

```python
pd.merge(staff_df, student_df, how='left', on = 'Name')
```

## Right Join
RIGHT (OUTER) JOIN: Returns all records from the right table, and the matched records from the left table
![](../img/img_rightjoin.gif)

```python
pd.merge(staff_df, student_df, how='right', on = 'Name')
```

```python
pd.merge(staff_df, student_df, how='right', on = 'Name')
```

```python
# rename student_df column name
student_df = student_df.reset_index(drop=False)
```

```python
student_df.rename(columns={'Name': 'student_name'}, inplace=True)
```

```python
student_df
```

```python
# error will be occured because 'Name' column is not common now
pd.merge(staff_df, student_df, how='right', on = 'Name')
```

```python
pd.merge(staff_df, student_df, how='right', left_on = 'Name', right_on = 'student_name')
```

## 4.9.2 Concatenating Dataframes

```python
# Creating two sample DataFrames
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
```

```python
df1
```

```python
df2 
```

### 4.9.1 Row concatenation (Vertical Concatenation)

```python
# Concatenating along rows
result = pd.concat([df1, df2], axis=0)
result
```

### 4.9.2 Column Concatenation (Horizontal Concatenation)

```python
# Concatenating along columns
result = pd.concat([df1, df2], axis=1)
result
```
